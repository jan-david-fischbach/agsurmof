```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
folder = "out/3poles/"
folder = "out/1pole/"
```

```python
results = np.load(folder+"ag_surmof.npy",allow_pickle=True).item(0)
```

```python
poles = results['filtered_poles']
thickness = results['thicknesses']
```

```python
to_THz = 300/2/np.pi
ylim = (300, 500)
```

```python
for t, p in zip(thickness, poles):
    plt.scatter([t]*len(p), np.real(p)*to_THz, color="k", marker="x")

#plt.ylim(5, 12)
plt.ylabel("$\Re\{f\}$ [Hz]")
plt.xlabel("Thickness [$\mu m$]")
plt.ylim(ylim)
```

```python
for t, p in zip(thickness, poles):
    plt.scatter([1/t]*len(p), np.real(p)*to_THz, color="k", marker="x")

plt.xlim(0, 8)
plt.ylabel("$\Re\{f\}$ [Hz]")
plt.xlabel("1/Thickness [$\mu m^{-1}$]")
plt.ylim(ylim)
```

```python
all_results = np.load(folder+"all_ag_surmof.npy",allow_pickle=True).item(0)
```

```python
all_poles = all_results['all_poles']
```

```python
for t, c_p in zip(thickness, all_poles):
    for p in c_p:
        plt.scatter([t]*len(p), np.real(p)*to_THz, color="k", marker="x", alpha=0.1)

plt.ylabel("$\Re\{f\}$ [Hz]")
plt.xlabel("Thickness [$\mu m$]")
plt.ylim(ylim)
```
